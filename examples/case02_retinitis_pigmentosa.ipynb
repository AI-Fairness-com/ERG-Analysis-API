The PRPF31 variant produces a rod-predominant dystrophy in early stages, and secondary cone involvement, while present (mildly reduced flicker amplitude, AMBER), has not yet produced a flicker implicit time delay."*

```python
print("\n" + "=" * 60)
print("CLINICAL INTERPRETATION: PRESERVED FLICKER TIMING")
print("=" * 60)
print("The preserved LA 30 Hz flicker implicit time (GREEN) is expected in")
print("early-stage RP. The SHAP attribution explains this: the RP classification")
print("is driven entirely by the rod-pathway features (DA 0.01 extinction,")
print("DA 3.0 b-wave reduction, b/a ratio), while the intact flicker implicit time")
print("provides a small negative contribution to the RP probability, modestly")
print("pulling toward a cone-predominant disease class.")
```

## Step 6: SHAP Attribution – Diffuse Negative Pattern

**Source:** Chapter 19 §19.3.3

> *"The RP SHAP pattern is qualitatively different from the DR pattern. Where DR showed a focused positive attribution concentrated in the OP time-frequency region, RP shows diffuse negative attribution across virtually the entire spectrogram: the model's classification is driven by the near-total absence of normal spectral energy rather than by the presence of a specific pathological pattern."*

```python
print("\n" + "=" * 60)
print("SPECTROGRAM SHAP SALIENCY PATTERN")
print("=" * 60)
print("Widespread BLUE (negative SHAP) across the 75–300 Hz OP band")
print("  → Near-total absence of OP spectral energy is the primary driver")
print("\nWidespread BLUE (negative SHAP) across the 0–30 Hz b-wave region")
print("  → Severe reduction of b-wave spectral energy confirms rod loss")
print("\nSmall positive saliency (RED) in pre-stimulus noise region")
print("  → Model attends to the absence of signal where it expects to find it")
print("\nPre-stimulus region near-zero saliency")
print("  → Classification is not driven by pre-stimulus artifact")
```
