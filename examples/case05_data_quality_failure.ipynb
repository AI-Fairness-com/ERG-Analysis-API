# Case Study 5: Data Quality Failure – Two Scenarios

## Patient Information

**Patient:** Mr. G., 71-year-old male

**Clinical context:** Known intermediate AMD, receiving 6-monthly ERG monitoring per Chapter 18 protocol. Skin electrode used at this visit (patient declined DTL fiber due to discomfort following a recent eyelid procedure).

**Two scenarios:** Scenario A (borderline signal quality – WARNING), Scenario B (missing metadata field – HALT)

**Source:** ERG Manuscript, Chapter 19 §19.9

## Scenario A: SIGNAL_QUALITY_WARNING

**Source:** Chapter 19 §19.9.1

**Condition:** LA 3.0 signal with SNR = 2.1 dB. Skin electrode thresholds: PASS ≥3.0 dB, WARNING 1.5–2.9 dB, FAIL < 1.5 dB. At SNR = 2.1 dB, recording is in the WARNING band.

```python
print("=" * 60)
print("SCENARIO A: SIGNAL_QUALITY_WARNING")
print("=" * 60)

print("\nSNR = 2.1 dB (skin electrode)")
print("PASS threshold: ≥3.0 dB")
print("WARNING range: 1.5–2.9 dB")
print("\nResult: WARNING flag assigned.")
```

### Layer 1 Report Header (WARNING Mode)

```python
print("\n" + "-" * 60)
print("LAYER 1 REPORT HEADER – WARNING MODE")
print("-" * 60)
print("\n⚠️ SIGNAL QUALITY WARNING")
print("\nThe SNR of this recording (2.1 dB) is in the warning range for")
print("skin electrode recordings (1.5–2.9 dB; PASS threshold: 3.0 dB).")
print("\nThe Traffic Light result and classification probabilities below are")
print("provided for reference only and must not be used as the basis for")
print("clinical decisions without specialist review.")
print("\nSHAP attribution is not displayed for low-confidence signal quality recordings.")
print("\nAction required: repeat recording with DTL fiber or gold foil electrode")
print("to obtain a clinically reliable result.")
```

### WARNING Mode Processing Behavior

```python
print("\n" + "-" * 60)
print("WARNING MODE PROCESSING BEHAVIOR")
print("-" * 60)
print("\n✅ Pipeline continues processing (does NOT halt)")
print("✅ Classification output is generated")
print("❌ SHAP attribution is SUPPRESSED (unreliable on low-SNR signals)")
print("✅ Layer 1 report displays grey LOCK RESULT UNRELIABLE badge")
print("✅ All report layers carry mandatory quality caveats")
print("\nRationale: A WARNING-level recording may still provide useful directional")
print("information (e.g., a clearly RED result increases clinical concern).")
print("A complete report with explicit caveats gives more information than no report.")
```

### Expected Classification (WARNING Mode)

```python
print("\n" + "-" * 60)
print("EXPECTED CLASSIFICATION (WARNING MODE – FOR REFERENCE ONLY)")
print("-" * 60)
print("\nStage 1: Abnormal (calibrated probability 0.61)")
print("Stage 2: AMD 0.49, Normal 0.34, Cone Dystrophy 0.12")
print("Composite Traffic Light: AMBER (displayed in grey)")
print("\n⚠️ These probabilities are displayed in greyed format and must NOT be")
print("   used for clinical decision-making without repeat recording.")
```

## Scenario B: ELECTRODE_UNKNOWN – Pipeline Halt

**Source:** Chapter 19 §19.9.2

```python
print("\n" + "=" * 60)
print("SCENARIO B: ELECTRODE_UNKNOWN – PIPELINE HALT")
print("=" * 60)
print("\nCondition: The CSV file submitted to the API has the electrode_type field blank.")
print("The CSV schema validation step (Chapter 10 §10.2, Appendix A) runs before any")
print("signal processing.")
```

### Layer 1 Error Report

```python
print("\n" + "-" * 60)
print("LAYER 1 ERROR REPORT – PROCESSING HALTED")
print("-" * 60)
print("\n❌ PROCESSING HALTED: The required field electrode_type is missing")
print("   from the submitted CSV file.")
print("\nThe electrode type is required for:")
print("   (1) signal quality threshold assignment")
print("   (2) normative reference limit selection")
print("   (3) cross-electrode transference validation")
print("\nProcessing cannot continue without this field. No classification has been attempted.")
print("\nAction required: resubmit the CSV file with the electrode_type field completed.")
print("Valid values: contact_lens, gold_foil, dtl_fiber, skin")
print("\nRefer to Appendix A for the complete CSV field specification.")
```

### HALT Mode Processing Behavior

```python
print("\n" + "-" * 60)
print("HALT MODE PROCESSING BEHAVIOR")
print("-" * 60)
print("\n❌ Pipeline halts at schema validation step")
print("❌ No feature extraction attempted")
print("❌ No classification attempted")
print("❌ No SHAP computation attempted")
print("✅ Specific, actionable error report returned")
print("✅ Error report stored in Layer 4 audit log with timestamp and patient ID")
print("\nThis is the safe failure by design principle: never produce an unreliable")
print("classification without explicit notification.")
```

## Comparison of Three Pipeline Output States

**Source:** Chapter 19 §19.9, Figure 19.7

```python
print("\n" + "=" * 60)
print("COMPARISON OF THREE PIPELINE OUTPUT STATES")
print("=" * 60)

states = [
    ("PASS", "Green badge, full SHAP, full confidence", "SNR ≥ threshold"),
    ("WARNING", "Grey badge, SHAP suppressed, mandatory caveats", "SNR in warning range"),
    ("HALT", "Error report, no classification, specific remediation", "Missing required field")
]

print(f"\n{'State':<12} {'Output':<50} {'Trigger':<30}")
print("-" * 95)
for state, output, trigger in states:
    print(f"{state:<12} {output:<50} {trigger:<30}")

print("\nThe pipeline is designed to fail safely and never produce an unreliable")
print("result without explicit notification.")
```

## Summary

**Key Takeaways:**
1. **WARNING state (Scenario A):** Pipeline continues; SHAP suppressed; mandatory caveats; grey badge
2. **HALT state (Scenario B):** Pipeline stops; no classification; specific remediation instruction
3. Both are safe failure modes – never produce unreliable result without explicit notification
4. All failures are logged in Layer 4 audit record with timestamp and patient ID

**Clinical Actions:**
- **WARNING:** Repeat recording with higher-sensitivity electrode (DTL fiber or gold foil)
- **HALT:** Resubmit with complete electrode_type field

**Lesson:** Data quality validation is the first and most important step in the pipeline.
