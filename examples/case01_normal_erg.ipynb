**Table 19.1 values.** Each row: protocol, parameter, value, and the zone printed in the book under the ≥60y and ≤35y limits.

```python
ROWS = [
    ('DA 0.01', 'b-wave amplitude',      190,  'GREEN', 'GREEN'),
    ('DA 0.01', 'b-wave implicit time',  98,   'GREEN', 'GREEN'),
    ('DA 3.0',  'a-wave amplitude',      95,   'GREEN', 'RED'),
    ('DA 3.0',  'a-wave implicit time',  16.4, 'GREEN', 'AMBER'),
    ('DA 3.0',  'b-wave amplitude',      205,  'GREEN', 'GREEN'),
    ('DA 3.0',  'b-wave implicit time',  52.1, 'GREEN', 'GREEN'),
    ('DA 10.0', 'a-wave amplitude',      140,  'GREEN', 'GREEN'),
    ('DA 10.0', 'a-wave implicit time',  13.1, 'GREEN', 'GREEN'),
    ('DA 10.0', 'b-wave amplitude',      240,  'GREEN', 'GREEN'),
    ('DA 10.0', 'b-wave implicit time',  55,   'GREEN', 'GREEN'),
    ('LA 3.0',  'a-wave amplitude',      20,   'GREEN', 'GREEN'),
    ('LA 3.0',  'a-wave implicit time',  14.2, 'GREEN', 'GREEN'),
    ('LA 3.0',  'b-wave amplitude',      58,   'GREEN', 'AMBER'),
    ('LA 3.0',  'b-wave implicit time',  29.6, 'GREEN', 'GREEN'),
    ('LA 30 Hz', 'flicker amplitude',    45,   'GREEN', 'AMBER'),
    ('LA 30 Hz', 'flicker implicit time', 27.8, 'GREEN', 'GREEN'),
    ('LA 3.0',  'PhNR amplitude',        18.4, 'Not zoned', 'Not zoned'),
]
```

**Her own stratum (≥60y).** Every zoned value is GREEN, so the composite Traffic Light is GREEN. The PhNR has no published limits and is shown as a measured value.

```python
table, signal = zone_table([r[:3] + (r[3],) for r in ROWS], age=67)
print(table.drop(columns='Match').to_string(index=False))
print('\nComposite Traffic Light:', signal)
check(table)
```

**The ≤35y stratum, for comparison.** Limits fall with age. Applied to a 67-year-old, the ≤35y limits flag this normal recording with one RED and three AMBER values, which is why the pipeline selects the stratum from the age at the visit (§19.1.2).

```python
table35, signal35 = zone_table([r[:3] + (r[4],) for r in ROWS], age=30)
print(table35.drop(columns='Match').to_string(index=False))
print('\nComposite Traffic Light:', signal35)
check(table35)
```

**Archiving the reference recording (§19.1.4).** At a later visit the DA 3.0 b-wave is compared with this one against the coefficient of repeatability (CoR) of Chapter 15, 0.216 log µV. A fall of more than 39% counts as real change.

```python
cor = 0.216
threshold = 205 * 10 ** -cor
print(f'A later DA 3.0 b-wave below {threshold:.0f} µV would exceed the CoR '
      f'(a fall of {100 * (1 - 10 ** -cor):.0f}%).')
```
