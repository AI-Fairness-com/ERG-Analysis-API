## Step 3: The Pre-Perimetric Glaucoma Use Case

**Source:** Chapter 19 §19.5.1

> *"Standard automated perimetry detects field loss only after an estimated 25–40% of retinal ganglion cells have been lost (Quigley et al., 1989). The photopic negative response (PhNR) reflects the summed activity of retinal ganglion cells. It is reduced in proportion to ganglion cell loss, often before the threshold for Humphrey VF detection is reached."*

```python
print("=" * 60)
print("PRE-PERIMETRIC GLAUCOMA – CLINICAL SIGNIFICANCE")
print("=" * 60)
print("\nOCT shows RNFL thinning (78 μm, 4th percentile) – structural ganglion cell loss")
print("Humphrey visual field: within normal limits – no functional deficit on perimetry")
print("ERG PhNR: reduced (Z = -2.8, AMBER) – functional ganglion cell deficit detected")
print("\nThe ERG detects functional impairment before perimetry threshold decline.")
print("This is the earliest detectable functional change in glaucoma.")
```

## Step 4: SHAP Attribution – PhNR as Dominant Contributor

**Source:** Chapter 19 §19.5.3 – Figure 19.5

```python
print("\n" + "=" * 60)
print("SHAP ATTRIBUTION – GLAUCOMA CLASSIFICATION")
print("=" * 60)

shap_contributors = [
    ('PhNR amplitude (µV)', 0.31, 'red', 'Reduced PhNR – dominant driver (41% of total attribution)'),
    ('OP2 implicit time (ms)', 0.12, 'red', 'Normal OP2 – evidence against DR'),
    ('DA 3.0 b-wave amplitude', 0.09, 'red', 'Normal b-wave – evidence against RP and CSNB'),
    ('LA 30 Hz flicker amplitude', 0.07, 'red', 'Intact flicker – consistent with ganglion cell-selective pathology')
]

for i, (feature, shap_val, color, interpretation) in enumerate(shap_contributors, 1):
    print(f"{i}. {feature}: SHAP = {shap_val:+.2f}")
    print(f"   {interpretation}\n")

print("Focused, PhNR-dominant attribution pattern – definitive validation that the classifier")
print("has learned the correct electrophysiological basis for Glaucoma classification.")
```
