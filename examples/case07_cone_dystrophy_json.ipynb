# Case Study 7: Cone Dystrophy

## Patient Information

**Patient:** Mr. A., 28-year-old male

**Clinical context:** Progressive photophobia and color vision loss since age 22. Color vision testing: Ishihara 2/38 OD and OS. OCT: foveal thinning with cone mosaic disruption on adaptive optics imaging. Referred from neuro-ophthalmology for electrophysiological diagnosis.

**Electrode type:** Skin electrode (recording center does not have contact lens facility)

**Protocols:** LA protocols primary; DA 3.0 for rod function confirmation

**Age stratum:** Baker three-strata: ≤35y

**Expected finding:** Selective photopic abnormality with intact scotopic responses, the defining pattern of cone dystrophy.

**Source:** ERG Manuscript, Chapter 19 §19.7

## Step 3: The Diagnostic Signature – RED Photopic, GREEN Scotopic

**Source:** Chapter 19 §19.7.3

```python
print("=" * 60)
print("DIAGNOSTIC SIGNATURE: RED PHOTOPIC, GREEN SCOTOPIC")
print("=" * 60)
print("\nThe defining feature of this case is the dissociation between:")
print("  - Severely abnormal photopic responses (LA 3.0 b-wave RED, LA 30 Hz RED)")
print("  - Completely normal scotopic responses (DA 3.0 and DA 0.01 GREEN)")
print("\nThis pattern is pathognomonic for a cone-selective retinal dystrophy.")
print("\nThe SHAP waterfall chart reflects this directly:")
print("  - Top three contributors are all photopic parameters")
print("  - Normal DA 3.0 b-wave provides negative evidence against RP/CSNB/DR")
```
