## Step 2: The Schubert-Bornschein Pattern

**Source:** Chapter 19 §19.8.1

> *"Complete CSNB is caused by mutations in genes encoding proteins at the ON-bipolar cell synapse (most commonly NYX for X-linked CSNB or TRPM1 for autosomal recessive). The pathophysiology is a failure of signal transmission from photoreceptors to ON-bipolar cells. The photoreceptors themselves are intact, producing a normal or near-normal a-wave. The ON-bipolar cells cannot respond, producing an absent or severely reduced b-wave. The result is the characteristic Schubert-Bornschein ERG: a negative-appearing waveform where the a-wave is larger than the b-wave, producing a b/a ratio below 1.0."*

```python
print("=" * 60)
print("SCHUBERT-BORNSCHEIN PATTERN – COMPLETE CSNB")
print("=" * 60)
print("\nON-bipolar cell synapse failure:")
print("  - Photoreceptors intact → normal a-wave")
print("  - ON-bipolar cells cannot respond → absent/reduced b-wave")
print("  - Result: b/a ratio < 1.0 (negative ERG)")
```
