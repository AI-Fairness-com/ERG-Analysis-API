# Case 8: Complete congenital stationary night blindness

*Hands-On Electroretinography in the Age of AI*, Chapter 19, §19.8.

Mr. J. is in the ≤35y age stratum. In complete CSNB, signal transmission from the photoreceptors to the ON-bipolar cells fails: the a-wave is preserved, the dark-adapted b-wave is markedly reduced, and the b/a ratio falls below 1.0 (the Schubert-Bornschein waveform). This notebook assigns the zones of Table 19.9 with the pipeline code, checks zones and printed limits against the book, and computes the b/a ratio.

The values are constructed for teaching; they are not patient data. Limits: Baker et al. (2025), silver thread, ≤35y.

**Table 19.9 values.** The DA 0.01 response is not recordable and is entered with the Chapter 11 sentinels, as in Case 3.

```python
ABSENT_AMP, ABSENT_TIME = -9999.0, 9999.0   # Chapter 11 sentinels
ROWS = [
    ('DA 0.01', 'b-wave amplitude',      ABSENT_AMP,  'RED',   (138, 297)),
    ('DA 0.01', 'b-wave implicit time',  ABSENT_TIME, 'RED',   (72, 99.5)),
    ('DA 3.0',  'a-wave amplitude',      150,  'GREEN', (119, 238)),
    ('DA 3.0',  'a-wave implicit time',  15,   'GREEN', (13.5, 16)),
    ('DA 3.0',  'b-wave amplitude',      95,   'RED',   (191, 404)),
    ('DA 3.0',  'b-wave implicit time',  54,   'GREEN', (45.5, 58.5)),
    ('DA 10.0', 'b-wave amplitude',      110,  'RED',   (202, 409)),
    ('DA 10.0', 'b-wave implicit time',  55,   'GREEN', (47, 59.5)),
    ('LA 3.0',  'a-wave amplitude',      24,   'GREEN', (12, 43)),
    ('LA 3.0',  'a-wave implicit time',  14.1, 'GREEN', (13, 15.5)),
    ('LA 3.0',  'b-wave amplitude',      60,   'AMBER', (73, 227)),
    ('LA 3.0',  'b-wave implicit time',  33,   'RED',   (27, 31.5)),
    ('LA 30 Hz', 'flicker amplitude',    80,   'GREEN', (50, 166)),
    ('LA 30 Hz', 'flicker implicit time', 26.5, 'GREEN', (24, 28)),
    ('DA 3.0',  'b/a ratio',             0.63, 'Not zoned'),
    ('LA 3.0',  'PhNR amplitude',        24.1, 'Not zoned'),
]
```

**Zones.** The dark-adapted b-waves are RED while the DA 3.0 a-wave is GREEN; the photopic b-wave is delayed (RED) and reduced (AMBER), and the 30 Hz flicker is normal.

```python
table, signal = zone_table(ROWS, age=25)
table['Value'] = table['Value'].replace({ABSENT_AMP: 'Not recordable',
                                         ABSENT_TIME: 'Not recordable'})
print(table.drop(columns='Match').to_string(index=False))
print('\nComposite Traffic Light:', signal)
check(table)
```

**The b/a ratio (§19.8.2).** The ratio has no published limits and is not zoned, but it carries the diagnosis: about 0.6 in complete CSNB against about 2.0 in normal eyes (Tremblay, Laroche and De Becker, 1995).

```python
a_amp, b_amp = 150, 95
ratio = b_amp / a_amp
print(f'DA 3.0 b/a ratio: {b_amp} / {a_amp} = {ratio:.2f}')
print('Schubert-Bornschein waveform (b/a below 1.0):', ratio < 1.0)
assert round(ratio, 2) == 0.63
```

**SHAP attribution (§19.8.3).** Constructed for teaching. The attribution rests on a few features, led by the b/a ratio.

```python
for name, v in [('b/a ratio (0.63)', 0.34), ('DA 3.0 b-wave amplitude (95 µV)', 0.28),
                ('DA 3.0 a-wave amplitude, preserved (150 µV)', 0.16),
                ('LA 30 Hz flicker, normal (80 µV)', 0.08)]:
    print(f'{name:44s} {v:+.2f}')
```
