## Step 3: Age-Stratification – The Case for Baker Three-Strata Correction

**Source:** Chapter 19 §19.1.2

```python
print("=" * 60)
print("AGE STRATIFICATION DEMONSTRATION")
print("=" * 60)

# Without age correction (using full 18-80y reference range)
print("\nWITHOUT age correction (full 18-80y reference range):")
print("  LA 30 Hz flicker amplitude: 62 μV (Z = -1.29) → GREEN")
print("  PhNR amplitude: 18.4 μV (Z = -1.60) → GREEN")

# With age correction (≥60y stratum)
print("\nWITH age correction (≥60y stratum, Baker three-strata):")
print("  LA 30 Hz flicker amplitude: 62 μV (Z = -0.88) → GREEN")
print("  PhNR amplitude: 18.4 μV (Z = -0.58) → GREEN")

print("\nThe teaching point:")
print("A different 67-year-old patient with a borderline PhNR of 11.2 μV would produce:")
print("  WITHOUT age correction: Z = -2.76 → AMBER (false positive)")
print("  WITH age correction: Z = -1.80 → GREEN (correct)")
print("\nAge correction prevents false AMBER results in healthy older patients.")
```
