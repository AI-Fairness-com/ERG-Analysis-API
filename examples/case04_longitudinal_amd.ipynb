## Step 5: Longitudinal Monitoring – Three Serial Recordings

**Source:** Chapter 18 §18.1.2; Chapter 19 §19.6 (Longitudinal component)

```python
print("\n" + "=" * 60)
print("LONGITUDINAL MONITORING – SERIAL RECORDINGS")
print("=" * 60)

# Simulated serial LA 30 Hz flicker amplitude data
visits = ['Baseline', '6 months', '12 months', '18 months']
flicker_amp = [44, 41, 38, 35]  # μV
cor_threshold = 7.7  # CoR for LA 30 Hz flicker (μV)

print("\nLA 30 Hz flicker amplitude over 4 visits:")
for i, (visit, amp) in enumerate(zip(visits, flicker_amp)):
    print(f"  {visit}: {amp} μV")

# Calculate changes
changes = []
for i in range(1, len(flicker_amp)):
    delta = flicker_amp[i] - flicker_amp[i-1]
    changes.append(delta)

print("\nVisit-to-visit changes:")
for i, delta in enumerate(changes, 1):
    flag = "SIGNIFICANT_PROGRESSION" if abs(delta) >= cor_threshold else "BELOW_MDC95"
    print(f"  {visits[i-1]} → {visits[i]}: {delta:+.0f} μV ({flag})")

print(f"\nCoefficient of Repeatability (CoR) threshold: {cor_threshold} μV")
```

## Step 6: Linear Mixed-Effects Model for Rate of Change

**Source:** Chapter 18 §18.4

```python
print("\n" + "=" * 60)
print("LINEAR MIXED-EFFECTS MODEL – RATE OF CHANGE")
print("=" * 60)

print("\nPopulation-level slope (estimated from AMD cohort): -1.8 μV/year")
print("Individual random slope (Mrs. H.): -1.2 μV/year")
print("Individual annual rate of change: -3.0 μV/year")
print("95% CI: [-4.2, -1.8] μV/year")
print("\nThis slope is significantly different from zero (p < 0.001).")
print("\nProgression flag triggered: negative trajectory confirmed.")
print("Estimated decline: 3.0 μV per year (95% CI: 1.8–4.2 μV per year).")
```
