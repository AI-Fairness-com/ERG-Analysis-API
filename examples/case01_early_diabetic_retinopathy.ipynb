# Case Study 1: Early Diabetic Retinopathy

## Patient Information

**Patient:** Mr. K., 52-year-old male

**Clinical context:** Type 2 diabetes mellitus for 9 years. HbA1c 71 mmol/mol (8.6%). Annual diabetic eye screening normal. No visible retinopathy on fundoscopy or color fundus photography at today's visit. Referred by diabetologist for baseline ERG given duration of diabetes and suboptimal glycemic control.

**Electrode type:** DTL fiber (bilateral)

**Protocols:** All five ISCEV standard protocols recorded

**Age stratum:** Baker three-strata: 36-59y

**Expected finding:** Subtle inner retinal dysfunction preceding visible retinopathy: OP2 implicit time delay and OP power ratio reduction, with normal b-wave amplitude – the electrophysiological signature of early diabetic maculopathy.

**Source:** Chapter 19 §19.2.1 – Table 19.2
